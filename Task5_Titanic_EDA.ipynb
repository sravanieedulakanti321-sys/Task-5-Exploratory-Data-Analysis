# Task 5: Exploratory Data Analysis (EDA)

**Dataset:** Titanic passenger dataset  
**Tools:** Python, Pandas, Matplotlib, Seaborn

## Objective
Explore the dataset using descriptive statistics and visualizations, identify patterns, trends, missing values, and unusual observations, and summarize practical insights.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
DATA_URL = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

# Prefer the included local CSV; download from the public source if it is not present.
if os.path.exists("titanic.csv"):
    df = pd.read_csv("titanic.csv")
else:
    df = pd.read_csv(DATA_URL)

print("Dataset shape:", df.shape)
df.head()
```

## 1. Data overview
Use `info()` to inspect data types and non-null values, `describe()` for numeric summaries, and `value_counts()` for categorical frequencies.

```python
df.info()
```

```python
display(df.describe(include="all").T)
```

```python
print("Survival counts:")
display(df["survived"].value_counts().rename(index={0: "Did not survive", 1: "Survived"}))
print("\nPassenger class counts:")
display(df["class"].value_counts())
print("\nSex counts:")
display(df["sex"].value_counts())
```

## 2. Missing values
Missing values may bias comparisons. Review the count and percentage before choosing a treatment. This notebook leaves the original data unchanged and drops missing values only for plots that require them.

```python
missing = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percent": (df.isna().mean() * 100).round(2)
}).sort_values("missing_count", ascending=False)
display(missing[missing["missing_count"] > 0])
```

## 3. Univariate analysis
Histograms show numeric distributions; boxplots help flag spread and potential outliers.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df["age"].dropna(), bins=25, kde=True, ax=axes[0])
axes[0].set_title("Age distribution")
sns.boxplot(x=df["fare"].dropna(), ax=axes[1])
axes[1].set_title("Fare spread and potential outliers")
plt.tight_layout()
plt.show()
```

**Observation guide:** Age values are concentrated among adults, while fares are typically right-skewed with a small number of high-fare observations. Confirm these patterns from the generated charts.

## 4. Bivariate analysis
Compare survival with sex and passenger class, and explore the relationship between age and fare.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.barplot(data=df, x="sex", y="survived", errorbar=None, ax=axes[0])
axes[0].set_title("Survival rate by sex")
axes[0].set_ylabel("Survival rate")
sns.barplot(data=df, x="class", y="survived", order=["First", "Second", "Third"], errorbar=None, ax=axes[1])
axes[1].set_title("Survival rate by passenger class")
axes[1].set_ylabel("Survival rate")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="age", y="fare", hue="survived", alpha=0.65)
plt.title("Age vs fare by survival outcome")
plt.tight_layout()
plt.show()
```

## 5. Multivariate analysis and correlation
The heatmap uses numeric columns only. Correlation describes association and does not prove causation.

```python
numeric = df.select_dtypes(include="number")
plt.figure(figsize=(9, 6))
sns.heatmap(numeric.corr(), annot=True, cmap="coolwarm", fmt=".2f", center=0)
plt.title("Correlation heatmap of numeric variables")
plt.tight_layout()
plt.show()
```

```python
pair_cols = [c for c in ["age", "fare", "sibsp", "parch", "survived"] if c in df.columns]
sns.pairplot(df[pair_cols].dropna().sample(min(300, len(df[pair_cols].dropna())), random_state=42),
             corner=True, diag_kind="hist")
plt.show()
```

## 6. Summary findings
- Survival outcomes are not evenly distributed; compare survivor and non-survivor counts.
- Survival rate varies by sex and passenger class in this dataset.
- Age and fare contain missing values; cabin/deck information is especially incomplete.
- Fare has a wide range and potential high-value outliers, so median and IQR can complement the mean and standard deviation.
- Correlations and plots reveal associations, not causal explanations.

## 7. Business-style recommendations
Although this is a historical passenger dataset, the analysis demonstrates useful data-analysis practices: prioritize data quality checks, segment outcomes by meaningful groups, investigate outliers before removing them, and explain every visualization in plain language.
